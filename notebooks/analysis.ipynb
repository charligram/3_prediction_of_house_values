## 4. Separación de datos

Anteriormente este paso se había hecho después, pero es MUY necesario hacerlo inmediatamente, para no generar un data leakage y que nuestras transformaciones dependan de datos que son parte de los test, es decir, no podemos sacar un promedio de tal columna para rellenar valores si ese promedio se hizo considerando datos que van a ser ocupados como test.

No importa si luego hacemos data cleaning y feature engineering en solo los datos que son para el entrenamiento, lo único que tenemos que hacer es aplicar la misma lógica dentro de los datos de prueba para que el modelo pueda funcionar. 

### 4.1 Separar el target

```python
X = initial_df.drop(columns=['SalePrice'])
y = initial_df['SalePrice']
```

### 4.2 Separar datos de entrenamiento y pruebas

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8, random_state=42)
```

## 5. Data cleaning y feature engineering

Ojo que aqui trabajaremos unicamente con los datos de entrenamiento, para saber que conversiones hacer y todo.

### 5.1 Identificar todas las columnas que su dato de NaN solo significa que no tiene, no dato faltante.

```python
valid_nan_cols = [
    'Alley', 'MasVnrType', 'BsmtQual', 'BsmtCond', 'BsmtExposure', 'BsmtFinType1', 'BsmtFinType2', 'FireplaceQu',
    'GarageType', 'GarageFinish', 'GarageQual', 'GarageCond', 'PoolQC', 'Fence', 'MiscFeature'
]
```

### 5.2 Rellenar con "None" esas columnas cuando haya un NaN

Primero aplicarlo a X_train.

```python
for col in valid_nan_cols:
    X_train[col] = X_train[col].fillna('None')
```

Luego aplicar lo mismo para X_test, para ir llevando un seguimiento y luego no tener df diferentes.

```python
for col in valid_nan_cols:
    X_test[col] = X_test[col].fillna('None')
```

Ahora luego de un gran relleno de valores podríamos ver que valores nos faltan que tienen nulos.

```python
print(X_train.isnull().sum().sort_values(ascending=False))
```

### 5.3 Trabajo con LotFrontage

Se podría rellenar de 0's, pero esto no tendría sentido, estaría diciendo que tiene distancia 0 hacia la calle, por ello, es mejor rellenar con la mediana.

Primero obtenemos la mediana, pero solo de los datos de X_train, ya que son los datos que sabemos.

```python
median_lotfrontage = X_train['LotFrontage'].median()
```

```python
X_train['LotFrontage'] = X_train['LotFrontage'].fillna(median_lotfrontage)
```

Ahora rellenar los valores de X_test, pero no con la mediana de sus valores, si no con la del entrenamiento para evitar leakage.

```python
X_test['LotFrontage'] = X_test['LotFrontage'].fillna(median_lotfrontage)
```

Ver nulos que quedan

```python
print(X_train.isnull().sum().sort_values(ascending=False))
```

### 5.4 Trabajo con GarageYrBlt

Aquí surgen 2 cosas, primero se podrían dejar los nulos en año 0, pero eso diría que se construyó en el año 0 lo cual sería ilógico, para ello se podría hacer una columna indicando si tiene o no garage "HasGarage".

Lo segundo, si ese registro que tiene nulo el GarageYrBlt puede tener algun dato dentro de GarageType, GarageQual, etc. Entonces también sería inconsistente que tenga un tipo pero luego tenga año 0 por lo tanto no tiene garage.

Así que primero veamos si dentro de las otras columnas hay algo que indique que si tenga garage tanto en X_train como en X_test.

```python
print(X_train.loc[
    ((X_train['GarageType'] != 'None') | (X_train['GarageFinish'] != 'None') | (X_train['GarageQual'] != 'None') | (X_train['GarageCond'] != 'None')) & (X_train['GarageYrBlt'].isnull()),
    ['GarageType', 'GarageFinish', 'GarageQual', 'GarageCond', 'GarageYrBlt']
])
```

```python
print(X_test.loc[
    ((X_test['GarageType'] != 'None') | (X_test['GarageFinish'] != 'None') | (X_test['GarageQual'] != 'None') | (X_test['GarageCond'] != 'None')) & (X_test['GarageYrBlt'].isnull()),
    ['GarageType', 'GarageFinish', 'GarageQual', 'GarageCond', 'GarageYrBlt']
])
```

Podemos ver que no tiene, asi que todas las columnas de GarageYrBlt que tienen NaN es porque no hay garage.

### 5.5 Rellenar información GarageYrBlt

Primero crearemos la columna HasGarage, si en GarageYrBlt hay un valor NO nulo, se deja un 1, si es nulo quedaría false, se pasa a int y eso es 0.

```python
X_train['HasGarage'] = (X_train['GarageYrBlt'].notnull()).astype(int)
```

En segundo lugar pasamos todos los NaN de GarageYrBlt en 0.

```python
X_train['GarageYrBlt'] = X_train['GarageYrBlt'].fillna(0)
```

### 5.6 Hacer lo mismo para X_test

HasGarage

```python
X_test['HasGarage'] = (X_test['GarageYrBlt'].notnull()).astype(int)
```

NaN a 0

```python
X_test['GarageYrBlt'] = X_test['GarageYrBlt'].fillna(0)
```
